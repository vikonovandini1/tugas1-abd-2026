```python
# ============================================================
# MILESTONE 3
# ANALISIS TEMPORAL, SPASIAL, DAN ADVANCED EDA
# DATASET: NATIONWIDE AGROCLIMATIC INDONESIA
# ============================================================

# 1. IMPORT LIBRARY
from pathlib import Path
import polars as pl
import duckdb
import pandas as pd
import numpy as np
import plotly.express as px
import plotly.graph_objects as go

print("=" * 60)
print("MILESTONE 3 - ANALISIS DATA AGROKLIMAT INDONESIA")
print("=" * 60)


# 2. MEMUAT DATASET HASIL CLEANING
PROJECT_ROOT = Path.cwd().resolve()

if PROJECT_ROOT.name == "notebooks":
    PROJECT_ROOT = PROJECT_ROOT.parent

DATA_PATH = (
    PROJECT_ROOT
    / "data"
    / "processed"
    / "agroclimate_bersih.parquet"
)

if not DATA_PATH.exists():
    raise FileNotFoundError(
        f"Dataset tidak ditemukan: {DATA_PATH}"
    )

df = pl.read_parquet(DATA_PATH)

print("\n========== INFORMASI DATASET ==========")
print("Jumlah baris:", f"{df.height:,}")
print("Jumlah kolom:", df.width)
print("Daftar kolom:", df.columns)

display(df.head())


# 3. ANALISIS TEMPORAL MENGGUNAKAN POLARS
print("\n========== ANALISIS TEMPORAL ==========")

df = df.sort("date")

df_temporal = (
    df.group_by_dynamic(
        "date",
        every="1mo"
    )
    .agg([
        pl.col("temperature_2m_mean").mean().alias("rata_suhu"),
        pl.col("temperature_2m_max").mean().alias("suhu_maksimum"),
        pl.col("temperature_2m_min").mean().alias("suhu_minimum"),
        pl.col("precipitation_sum").mean().alias("rata_curah_hujan"),
        pl.col("relative_humidity_2m_mean").mean().alias("rata_kelembapan"),
        pl.col("water_balance").mean().alias("rata_water_balance")
    ])
    .sort("date")
)

# Moving average 3 bulan
df_temporal = df_temporal.with_columns(
    pl.col("rata_suhu")
    .rolling_mean(window_size=3)
    .alias("moving_average_suhu_3_bulan")
)

display(df_temporal.head(12))


# 4. MENYIAPKAN DATA VISUALISASI
print("\n========== PERSIAPAN VISUALISASI ==========")

data_temporal = df_temporal.to_pandas()

df_sample = (
    df.sample(n=min(20000, df.height), seed=42)
    .to_pandas()
)

print("Jumlah data sampel:", len(df_sample))


# 5. VISUALISASI 1: TREN SUHU BERDASARKAN WAKTU
print("\nVisualisasi 1: Tren Suhu")

fig1 = go.Figure()

fig1.add_trace(go.Scatter(
    x=data_temporal["date"],
    y=data_temporal["rata_suhu"],
    mode="lines",
    name="Rata-rata Suhu"
))

fig1.add_trace(go.Scatter(
    x=data_temporal["date"],
    y=data_temporal["moving_average_suhu_3_bulan"],
    mode="lines",
    name="Moving Average 3 Bulan"
))

fig1.update_layout(
    title="Visualisasi 1: Tren Suhu Rata-rata Indonesia",
    xaxis_title="Tanggal",
    yaxis_title="Suhu",
    hovermode="x unified",
    xaxis=dict(
        rangeslider=dict(visible=True)
    )
)

fig1.show()


# 6. VISUALISASI 2: DISTRIBUSI SUHU
print("\nVisualisasi 2: Distribusi Suhu")

fig2 = px.box(
    df_sample,
    y="temperature_2m_mean",
    points="outliers",
    title="Visualisasi 2: Distribusi Suhu Rata-rata",
    labels={
        "temperature_2m_mean": "Suhu Rata-rata"
    }
)

fig2.show()


# 7. VISUALISASI 3: KORELASI ANTARVARIABEL
print("\nVisualisasi 3: Korelasi Antarvariabel")

kolom_korelasi = [
    "temperature_2m_mean",
    "temperature_2m_max",
    "temperature_2m_min",
    "precipitation_sum",
    "relative_humidity_2m_mean",
    "et0_fao_evapotranspiration",
    "rainfall_30d",
    "water_balance"
]

data_korelasi = (
    df.select(kolom_korelasi)
    .sample(n=min(20000, df.height), seed=42)
    .to_pandas()
)

matriks_korelasi = data_korelasi.corr()

fig3 = px.imshow(
    matriks_korelasi,
    text_auto=".2f",
    color_continuous_scale="RdBu_r",
    zmin=-1,
    zmax=1,
    title="Visualisasi 3: Heatmap Korelasi Variabel Iklim"
)

fig3.show()


# 8. VISUALISASI 4: DISTRIBUSI QUALITY CONTROL
print("\nVisualisasi 4: Quality Control")

qc_data = (
    df.group_by("qc_range_flag")
    .len()
    .sort("qc_range_flag")
    .with_columns(
        pl.col("qc_range_flag")
        .cast(pl.String)
        .alias("status_qc")
    )
    .to_pandas()
)

fig4 = px.bar(
    qc_data,
    x="status_qc",
    y="len",
    text="len",
    title="Visualisasi 4: Distribusi Quality Control Data",
    labels={
        "status_qc": "Status QC",
        "len": "Jumlah Data"
    }
)

fig4.show()


# 9. VISUALISASI 5: POLA MUSIMAN CURAH HUJAN
print("\nVisualisasi 5: Pola Musiman Curah Hujan")

df_musiman = (
    df.with_columns([
        pl.col("date").dt.year().alias("tahun"),
        pl.col("date").dt.month().alias("bulan")
    ])
    .group_by(["tahun", "bulan"])
    .agg(
        pl.col("precipitation_sum")
        .mean()
        .alias("rata_curah_hujan")
    )
    .sort(["tahun", "bulan"])
)

data_musiman = df_musiman.to_pandas()

tabel_musiman = data_musiman.pivot(
    index="tahun",
    columns="bulan",
    values="rata_curah_hujan"
)

fig5 = px.imshow(
    tabel_musiman,
    text_auto=".2f",
    aspect="auto",
    color_continuous_scale="Blues",
    title="Visualisasi 5: Pola Curah Hujan Bulanan dan Tahunan",
    labels={
        "x": "Bulan",
        "y": "Tahun",
        "color": "Rata-rata Curah Hujan"
    }
)

fig5.show()


# 10. VISUALISASI 6: ANALISIS SPASIAL
print("\nVisualisasi 6: Persebaran Suhu Indonesia")

df_spasial = (
    df.group_by(["grid_id", "latitude", "longitude"])
    .agg([
        pl.col("temperature_2m_mean")
        .mean()
        .alias("rata_suhu"),

        pl.col("precipitation_sum")
        .mean()
        .alias("rata_curah_hujan")
    ])
)

if df_spasial.height > 3000:
    df_spasial = df_spasial.sample(n=3000, seed=42)

data_spasial = df_spasial.to_pandas()

fig6 = px.scatter_geo(
    data_spasial,
    lat="latitude",
    lon="longitude",
    color="rata_suhu",
    hover_name="grid_id",
    hover_data=["rata_curah_hujan"],
    color_continuous_scale="RdYlBu_r",
    projection="natural earth",
    title="Visualisasi 6: Persebaran Suhu Rata-rata di Indonesia"
)

fig6.update_geos(
    lataxis_range=[-12, 7],
    lonaxis_range=[94, 142],
    showcountries=True,
    showcoastlines=True
)

fig6.show()


# 11. RINGKASAN HASIL ANALISIS
print("\n" + "=" * 60)
print("RINGKASAN HASIL ANALISIS")
print("=" * 60)

# Periode suhu tertinggi
bulan_terpanas = data_temporal.loc[
    data_temporal["rata_suhu"].idxmax()
]

print("\n1. Periode dengan rata-rata suhu tertinggi:")
print(bulan_terpanas[["date", "rata_suhu"]])

# Periode suhu terendah
bulan_terdingin = data_temporal.loc[
    data_temporal["rata_suhu"].idxmin()
]

print("\n2. Periode dengan rata-rata suhu terendah:")
print(bulan_terdingin[["date", "rata_suhu"]])

# Periode curah hujan tertinggi
bulan_hujan = data_temporal.loc[
    data_temporal["rata_curah_hujan"].idxmax()
]

print("\n3. Periode dengan rata-rata curah hujan tertinggi:")
print(bulan_hujan[["date", "rata_curah_hujan"]])

# Korelasi terkuat
korelasi_abs = matriks_korelasi.abs().copy(deep=True)

nilai_korelasi_abs = korelasi_abs.to_numpy(copy=True)

np.fill_diagonal(nilai_korelasi_abs, np.nan)

korelasi_abs = pd.DataFrame(
    nilai_korelasi_abs,
    index=matriks_korelasi.index,
    columns=matriks_korelasi.columns
)

pasangan_terkuat = korelasi_abs.stack().idxmax()

nilai_korelasi = matriks_korelasi.loc[
    pasangan_terkuat[0],
    pasangan_terkuat[1]
]

print("\n4. Pasangan variabel dengan korelasi terkuat:")
print(pasangan_terkuat)
print("Nilai korelasi:", round(nilai_korelasi, 3))

# Grid dengan suhu tertinggi
grid_terpanas = (
    df_spasial
    .sort("rata_suhu", descending=True)
    .head(1)
)

print("\n5. Grid dengan rata-rata suhu tertinggi:")
display(grid_terpanas)

# Distribusi QC
print("\n6. Distribusi Quality Control:")
display(qc_data)


# 12. KESIMPULAN SEMENTARA
print("\n" + "=" * 60)
print("KESIMPULAN SEMENTARA MILESTONE 3")
print("=" * 60)

print("""
1. Analisis temporal dilakukan menggunakan Polars.
2. Moving average suhu 3 bulan berhasil dihitung.
3. Enam visualisasi interaktif dibuat menggunakan Plotly.
4. Distribusi dan outlier suhu dianalisis.
5. Korelasi antarvariabel iklim dianalisis.
6. Pola curah hujan berdasarkan bulan dan tahun dianalisis.
7. Persebaran suhu berdasarkan koordinat geografis divisualisasikan.

Analisis ini menjadi dasar untuk menyusun minimal 5 insight
berdasarkan hasil dataset.
""")

print("\nMILESTONE 3 SELESAI DIJALANKAN.")
```
